### Machine Learning

The algorithm learns the patterns from examples.

```text
Data + Correct Answers
          ↓
   Learning Algorithm
          ↓
      Learned Model
          ↓
   Prediction on New Data
```

---

## 2. Why Do We Use Machine Learning?

Some problems are difficult to solve by writing explicit rules.

For example, consider detecting whether an image contains a **cat**.

There can be:

* Different cat breeds
* Different poses
* Different lighting
* Different backgrounds
* Partially hidden cats
* Different camera angles

Writing rules for every possible situation would be difficult.

Instead, we can provide many examples:

```text
Cat images
Dog images
```

The ML algorithm learns useful patterns from these examples.

> **Machine Learning is useful when it is difficult to manually program all the rules needed to solve a problem.**

# 4. Supervised Learning

## What is Supervised Learning?

**Supervised learning** is a type of machine learning where an algorithm learns from **training examples with known correct answers**.

The algorithm is given:

```text
Input (x) + Correct Output (y)
             ↓
       ML Algorithm
             ↓
       Learned Model
```

The model then uses what it learned to make predictions on **new inputs**.

> **Supervised learning = learning from examples with known/correct answers.**

---

# 5. What is a Label?

A **label** is the **correct output/answer associated with a training example**.

### Example — Spam Detection

| Email             | Label    |
| ----------------- | -------- |
| "Win free money!" | Spam     |
| "Meeting at 3 PM" | Not Spam |

Here:

```text
Input = Email
Label = Spam / Not Spam
```

### Example — House Price

| House Size | Label |
| ---------: | ----: |
|   1000 ft² | $180k |
|   1250 ft² | $220k |
|   1500 ft² | $250k |

Here:

```text
Input = House Size
Label = Actual House Price
```

Therefore:

> **Label = the correct target/output the model is learning to predict.**

A label can be **text or a number**.

---

# 6. Types of Supervised Learning

Supervised learning mainly includes two important types:

```text
Supervised Learning
        |
        +--------------------+
        |                    |
    Regression          Classification
        |                    |
 Predict quantities      Predict categories
```

# 7. Regression

## What is Regression?

**Regression** is a supervised learning task where the model predicts a **numerical quantity**.

### Examples

```text
House price  → ₹90 lakh
Temperature  → 31.5°C
Energy usage → 4.7 kWh
Salary       → ₹8.5 lakh
Distance     → 12.7 km
```

These numbers represent **amounts or measurable values**.

> **Regression → predicts a numerical quantity.**

---

# 8. What is a Numerical Quantity?

A **numerical quantity** is a number that represents an **amount or measurable value**.

|    Value | Represents  |
| -------: | ----------- |
|    56 kg | Weight      |
|   165 cm | Height      |
|   31.5°C | Temperature |
| ₹90 lakh | Price       |
|  4.7 kWh | Energy      |
|  12.7 km | Distance    |

For regression, the output number has a real quantitative meaning.

---

# 9. House-Price Example

Suppose we have the following training data:

| House Size (ft²) | Actual Price ($1000s) |
| ---------------: | --------------------: |
|             1000 |                   180 |
|             1250 |                   220 |
|             1500 |                   250 |
|             2000 |                   350 |
|             2500 |                   500 |

Here:

```text
Input / Feature x = House Size
Output / Target y = House Price
```

The training data contains the **correct answers**.

Therefore:

```text
Supervised Learning
        ↓
Regression
```

Why?

Because:

* The correct answers are provided → **Supervised Learning**
* The output is a numerical quantity → **Regression**

# 10. Linear Regression

The blue line in the house-price graph represents a **linear regression model**.

It learns a relationship between:

```text
x = House Size
y = House Price
```

The model can then take a **new house size** and predict its price.

### Example

```text
x = 1250 ft²
```

The model might predict:

```text
ŷ ≈ 220
```

where:

```text
ŷ = predicted output
```

Because the graph's y-axis is in **$1000s**:

```text
220 = $220,000
```

---

# 11. Linear Regression Model

A simple linear regression model can be written as:

[
f_{w,b}(x) = wx + b
]

Where:

* `x` → input / feature
* `w` → weight / parameter
* `b` → bias / parameter
* `f_{w,b}(x)` → predicted output
* `ŷ` → predicted output

Therefore:

[
\hat{y} = f_{w,b}(x)
]

The model learns suitable values for `w` and `b` so that its predictions are close to the actual training outputs.

# 15. Classification

## What is Classification?

**Classification** is a supervised learning task where the model predicts a **category or class**.

### Examples

```text
Email       → Spam / Not Spam

Image       → Cat / Dog

Transaction → Fraud / Not Fraud

Patient     → Disease / No Disease
```

The model is choosing **which category** an example belongs to.

> **Classification → predicts a category/class.**

---

# 16. What is a Category?

A **category** is a class or group to which an example belongs.

Examples:

```text
Cat
Dog
Spam
Not Spam
Fraud
Not Fraud
Disease
No Disease
```

The model is answering:

> **Which category does this example belong to?**

---

# 17. Regression vs Classification

|                  | Regression            | Classification          |
| ---------------- | --------------------- | ----------------------- |
| Predicts         | Numerical quantity    | Category / Class        |
| Possible outputs | Many numerical values | Small set of categories |
| Example          | ₹90 lakh              | Spam                    |
| Example          | 31.5°C                | Cat                     |
| Example          | 4.7 kWh               | Fraud                   |
| Main question    | **How much?**         | **Which category?**     |

### Easy way to remember

```text
Regression → How much?
Classification → Which category?
```

---

# 18. Important: A Number Can Still Represent a Category

Suppose:

```text
0 = Cat
1 = Dog
```

If the model predicts:

```text
1
```

this is still **classification**.

Why?

Because `1` is being used as a **category code**.

It means:

```text
1 → Dog
```

It does **not** mean:

```text
1 unit of Dog
```

Therefore:

> **Look at what the output represents, not just whether it is written as a number.**

### Compare

```text
72.5 kg
```

`72.5` represents a measurable quantity.

→ **Regression**

```text
1 = Dog
```

`1` represents a category.

→ **Classification**

---

# 19. Why Regression "Predicts a Number"

When we say:

> **Regression → Predict a number**

we mean the output is a **numerical quantity**.

Examples:

```text
5
72.5
-3.2
0.00042
₹90,00,000
```

The important point is that the number represents a **quantity**.

It is not simply a code for a category.

---

# 20. Infinitely Many Possible Outputs in Regression

Regression can produce a very large or continuous range of numerical values.

For example, house prices could be:

```text
₹50,00,000
₹50,00,001
₹50,00,002
₹50,00,003
...
```

There can be an enormous or effectively continuous range of possible numerical outputs.

Therefore:

> **Regression → numerical values → potentially infinitely many possible outputs**

---

# 21. Small Number of Possible Outputs in Classification

Classification normally predicts one category from a **small set of possible classes**.

Example:

```text
Cat
Dog
```

Two possible classes.

Another example:

```text
Red
Green
Blue
```

Three possible classes.

Therefore:

> **Classification → categories → small number of possible outputs**

---

# 22. Supervised Learning is the Bigger Concept

Do **not** think:

```text
Regression = Supervised Learning
```

Instead:

```text
Supervised Learning
        |
        +--------------------+
        |                    |
    Regression          Classification
        |                    |
 Predict quantities      Predict categories
```

Therefore:

> **Regression is one type of supervised learning.**

> **Classification is another type of supervised learning.**
