# 🔥 LEARNING METHOD

For every important concept:

```text
Learn
  ↓
Understand
  ↓
Solve
  ↓
Implement From Scratch
  ↓
Use Library
  ↓
Build
  ↓
Explain
```

### Measure progress by:

- Can I explain the concept?
- Can I solve a small problem?
- Can I explain the mathematics?
- Can I implement it?
- Can I use the library?
- Can I build something with it?
- Can I explain it in an interview?
